Numberical Analysis using Numpy

```python
import pandas as pd
import numpy as np

df=pd.read_csv('/content/Student_Performance.csv')
scored=np.array(df[['math_score','science_score','english_score']])
print(scored)
```

```python
math_scored=np.array(df['math_score'])
print(math_scored)
science_scored=np.array(df['science_score'])
print(science_scored)
english_scored=np.array(df['english_score'])
print(english_scored)
```

```python
print(np.mean(math_scored))
print(np.mean(science_scored))
print(np.mean(english_scored))
```

```python
print(np.median(math_scored))
print(np.median(science_scored))
print(np.median(english_scored))
```

```python
print(np.std(math_scored))
print(np.std(science_scored))
print(np.std(english_scored))
```

Data Visualization using matplotlib

```python
import matplotlib.pyplot as plt
plt.hist(math_scored, bins=10)
plt.title('Distribution of Math Scores')
plt.xlabel('Math Score')
plt.ylabel('Frequency')
plt.show()
```

```python
import matplotlib.pyplot as plt

math_avg=np.mean(math_scored)
science_avg=np.mean(science_scored)
english_avg=np.mean(english_scored)

subjects = ['Math', 'Science', 'English']
average_scores = [math_avg, science_avg, english_avg]

plt.figure(figsize=(8, 6))
plt.bar(subjects, average_scores, color=['skyblue', 'lightcoral', 'lightgreen'])
plt.title('Average Scores by Subject')
plt.xlabel('Subject')
plt.ylabel('Average Score')
plt.ylim(0, 100) # Assuming scores are out of 100
plt.show()
```

```python
import matplotlib.pyplot as plt

plt.scatter(math_scored, science_scored, label='Math vs. Science', color='skyblue')
```

Group-wise Analysis using Pandas and Matplotlib

```python
import matplotlib.pyplot as plt
import pandas as pd
gender_scores = df.groupby('gender')[['math_score', 'science_score', 'english_score']].mean()
gender_scores.plot(kind='bar', figsize=(10, 6))
plt.title('Scores by Gender')
plt.ylabel('Average Score')
plt.xlabel('Gender')
plt.legend(title='Subject')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import pandas as pd
parental_eduation=pd.crosstab(df['parent_education'],df['extra_activities'])
plt.bar(parental_eduation.index,parental_eduation['yes'],color='skyblue')
plt.bar(parental_eduation.index,parental_eduation['no'],bottom=parental_eduation['yes'], color='lightcoral')
plt.title('Parental Education vs. Presence of Extra Activities')
plt.xlabel('Parental Education')
plt.ylabel('Count')
plt.show()
```

```python
import matplotlib.pyplot as plt
avg_scores_by_parent_education = df.groupby('parent_education')[['math_score', 'science_score', 'english_score']].mean()
avg_scores_by_parent_education.plot(kind='bar', figsize=(12, 7))
plt.title('Average Scores by Parental Education')
plt.xlabel('Parental Education')
plt.ylabel('Average Score')
plt.legend(title='Subject')
plt.xticks(rotation=45, ha='right') # Rotate x-axis labels for better readability
plt.tight_layout()
plt.show()
```
